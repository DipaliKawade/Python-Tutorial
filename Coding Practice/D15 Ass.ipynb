2. Product Class – __eq__()
Create a Product class with name and price. Implement __eq__() to check whether two
products have the same price.

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __eq__(self, other):
        return self.price == other.price


pd1 = Product("Pen", 20)
pd2 = Product("Book", 20)

print(pd1 == pd2)
```

3. Shopping Cart – __len__()
Create a ShoppingCart class containing a list of products. Implement __len__() so that
len(cart) returns the number of products in the cart.

```python
class ShoppingCart:
    def __init__(self, products):
        self.products = products

    def __len__(self):
        return len(self.products)


cart = ShoppingCart(["Pen", "Book", "Bag"])

print(len(cart))
```

4. Bank Account – __str__()
Create a BankAccount class with account holder name, account number, and balance.
Implement __str__() to display account information when the object is printed.


```python
class BankAccount:
    def __init__(self, name, account_no, balance):
        self.name = name
        self.account_no = account_no
        self.balance = balance

    def __str__(self):
        return f"Name: {self.name}, Account No: {self.account_no}, Balance: {self.balance}"


account = BankAccount("Dipali", 12345, 50000)

print(account)
```

5. Number Class – __add__()
Create a Number class that stores one number. Implement __add__() so that two
objects can be added


```python
class Number:
    def __init__(self, num):
        self.num = num

    def __add__(self, other):
        return self.num + other.num


num1 = Number(10)
num2 = Number(20)

print(num1 + num2)
```

6. Create a Number class and implement:
__add__()
__sub__()
__mul__()
__truediv__()
__mod__()

```python
class Number:
    def __init__(self, num):
        self.num = num

    def __add__(self, other):
        return self.num + other.num

    def __sub__(self, other):
        return self.num - other.num

    def __mul__(self, other):
        return self.num * other.num

    def __truediv__(self, other):
        return self.num / other.num

    def __mod__(self, other):
        return self.num % other.num


num1 = Number(10)
num2 = Number(3)

print("Addition:", num1 + num2)
print("Subtraction:", num1 - num2)
print("Multiplication:", num1 * num2)
print("Division:", num1 / num2)
print("Modulus:", num1 % num2)
```
