```python
class Parent:
    def __init__(self):
        print("this is parent class constructor")
    def show(self):
        print("this is the parent class")

class Child(Parent):
    def __init__(self):
        super(). __init__() 
        print("this is child class constructor")
    def display(self):
        print("this is the child class")

obj=Child()
# obj.show()
# obj.display()
```
