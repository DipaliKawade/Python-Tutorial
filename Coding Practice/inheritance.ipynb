# types of inheritance

# single level

```python
class Grandparent:
    def temp(self):
        print("this is grandparent class ")

class Parent(Grandparent):
    def display(self):
        print("this is parent class ")

obj=Parent()
obj.temp()
```

# multilevel

```python
class Grandparent:
    def temp(self):
        print("this is grandparent class ")

class Parent(Grandparent):
    def display(self):
        print("this is parent class ")

class Son(Parent):
    def data(self):
        print("this is son class ")
obj=Son()
obj.temp()
obj.display()
obj.data()




```

# multiple

```python
class father:
    def temp(self):
        print("this is father class ")

class mother:
    def display(self):
        print("this is mother class ")

class Son(father,mother):
    def data(self):
        print("this is son class ")
obj=Son()
obj.temp()
obj.display()
obj.data()




```
