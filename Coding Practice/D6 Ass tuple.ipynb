```python
# Q1 Write a program to find the largest and smallest element in a tuple without using max() and min(). 
tupleOne = (45, 12, 67, 10, 54)

sortedTuple = sorted(tupleOne)

print("Smallest =", sortedTuple[0])
print("Largest =", sortedTuple[-1])
```

```python
# Q2 Write a program to count the number of even and odd numbers in a tuple. 
tupleOne = (45, 12, 67, 10, 54)
even=0
odd=0
for i in tupleOne:
    if i%2 == 0:
        even= even+1
        print(i,"Even")
    else:
        odd= odd+1
        print(i,"odd")
print("Even number:", even)
print("Odd number:", odd)
```

```python
# Q3 Write a program to find all common elements between two tuples. 
tupleOne = (10, 20, 30, 40)
tupleTwo = (20, 40, 60)

Common = ()

for i in tupleOne:
    if i in tupleTwo:
        Common += (i,)

print(Common)
```

```python
# Q4 Write a program to create a new tuple containing only positive numbers from a given tuple
tupleOne = (10, 20, 30, 40, -20)

positive=()

for i in tupleOne:
    if i > 0:
        positive +=(i,)
print(positive)
```

```python
# Q5  Write a program to remove duplicate elements from a tuple while preserving the original order. 

tupleOne = (10, 20, 30, 40, 20, 10)

num=()

for i in tupleOne:
    if i not in num:
        num+=(i,)
print(num)
```

```python
# Q6 Write a program to separate all even and odd elements of a tuple into two different tuples. 
tupleOne =(10,15,20,25,30) 
even=()
odd=()

for i in tupleOne:
    if i%2==0:
        even +=(i,)
    else:
        odd +=(i,)
    
print("even list:",even)
print("odd list:",odd)
```

```python
# Q7 Write a program to find the intersection of two tuples.
tupleOne = (10, 20, 30, 40)
tupleTwo = (20, 40, 60, 10)
common=()


for i in tupleOne:
    for j  in tupleTwo:
        if i==j:
            common += (i,)

print(common, end=" ")
```

```python
# Q8 Write a program to find the union of two tuples (unique elements only).
tupleOne = (10, 20, 30, 40)
tupleTwo = (10, 20, 60, 40)

unique= ()
for i in tupleOne:
    if i not in tupleTwo:
        unique+=(i,)

for i in tupleTwo:
    if i not in tupleOne:
        unique += (i,)
print(unique)
```
