PROPERTIES OF DICTIONARY

```python
DictOne= {1:"dipali",2:"divya",3:"shubh"}   #mutable
print(DictOne)

DictOne[4]="shwetali"
print(DictOne)

DictOne[1]="Swati"
print(DictOne)

#if value is diff but the key is same then take the recently key value
DictOne= {1:"dipali",2:"divya",1:"shubh"}     #ordered
print(DictOne)                                  

DictOne= {1:"dipali",2:"divya",3:"shubh",4:"dipali"}     #if value is same but key is diff
print(DictOne)                                              #same valuue is  allowed

```

```python
#can store diff data types  #heterogeneous------------diff data type is used
DictOne= {1:"dipali",2:"divya",3:"True",4:"6.4"} 
print(DictOne)

DictOne= {1:1,2:2,3:3,4:4}    #homogeneous-----------same data type is used
print(DictOne)

DictOne[5]=5                #dynamic-------------size can be inc or dec
print(DictOne)

```
