```python
# Q1 Count Even and Odd Numbers in a List
listOne = [10, 15, 20, 25, 30, 35]

even = 0
odd = 0

for i in listOne:
    if i % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1

print("Even Numbers =", even)
print("Odd Numbers =", odd)
```

```python
# Q2 Create a New List Containing Only Positive Numbers
listOne = [-10, 20, -5, 40, 0, 15]

positive = []

for i in listOne:
    if i > 0:
        positive.append(i)

print(positive)
```

```python
# Q3 Write a program to remove all duplicate elements from a list while preserving the original order. 
listOne = [10, 20, 30, 20, 40, 10, 50]

num = []

for i in listOne:
    if i not in num:
        num.append(i)

print(num)
```

```python
# Q7 Write a program to reverse a list without using the reverse() method or slicing ([::-1]). 
listOne = [10, 20, 30, 40, 50]

new = []

for i in listOne:
    new = [i] + new

print(new)
```

```python
# Q8 Write a program to check whether a list is a Palindrome. 
listOne = [1, 2, 3, 2,1]

new = []

for i in listOne:
    new = [i] + new

if listOne == new:
    print("Palindrome")
else:
    print("Not Palindrome")

# OR another way

listOne = [1, 2, 3, 2, 1]
reverse =list(reversed(listOne))

if listOne == reverse:

    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# Q9 Find Common Elements Between Two Lists
listOne = [10, 20, 30, 40]
listTwo = [20, 40, 60]

Common = []

for i in listOne:
    if i in listTwo:
        Common.append(i)

print(Common)
```

```python
# Q10 Write a program to separate all even and odd numbers into two different lists.
listOne = [10, 15, 20, 25, 30]

even = []
odd = []

for i in listOne:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)

print("Even =", even)
print("Odd =", odd)
```
