```python
class BankAtm:

    def __init__(self):
        self.pin = " "
        self.balance = 0

    def Menu(self):

        while True:

            print("\n1. Enter 1 to set PIN")
            print("2. Enter 2 to check balance")
            print("3. Enter 3 to deposit money")
            print("4. Enter 4 to withdraw money")
            print("5. Enter 5 to change PIN")
            print("6. Enter 6 to exit")

            choice = int(input("\nEnter your choice: "))

            print("You selected:", choice)

            match choice:

                case 1:
                    self.set_pin()

                case 2:
                    self.check_balance()

                case 3:
                    self.deposit_money()

                case 4:
                    self.withdraw_money()

                case 5:
                    self.change_pin()

                case 6:
                    print("Thank you!")
                    break

                case _:
                    print("Invalid choice")

    def set_pin(self):

        if self.pin == " ":
            input_pin = input("Enter PIN: ")
            self.pin = input_pin
            print("PIN set successfully")
        else:
            print("PIN is already set")

    def check_balance(self):

        input_pin = input("Enter PIN: ")

        if input_pin == self.pin:
            print("Your balance is:", self.balance)
        else:
            print("Incorrect PIN")

    def deposit_money(self):

        input_pin = input("Enter PIN: ")

        if input_pin == self.pin:
            amount = int(input("Enter amount to deposit: "))

            self.balance = self.balance + amount

            print("Amount deposited successfully")
            print("Your balance is:", self.balance)
        else:
            print("Incorrect PIN")

    def withdraw_money(self):

        input_pin = input("Enter PIN: ")

        if input_pin == self.pin:
            amount = int(input("Enter amount to withdraw: "))

            if amount <= self.balance:
                self.balance = self.balance - amount

                print("Please collect your cash")
                print("Your balance is:", self.balance)
            else:
                print("Insufficient balance")
        else:
            print("Incorrect PIN")

    def change_pin(self):

        old_pin = input("Enter old PIN: ")

        if old_pin == self.pin:
            new_pin = input("Enter new PIN: ")
            self.pin = new_pin
            print("PIN changed successfully")
        else:
            print("Incorrect PIN")


obj1 = BankAtm()
obj1.Menu()
```

1. Create a Student Management System using Class and Object in Python.
What to Do
1. Create a class named Student.
2. Create a constructor __init__() to initialize:
o Student name
o Roll number
o Age
o Marks of 3 subjects
3. Create a display_details() method to display all student information.
4. Create a calculate_total() method to calculate the total marks.
5. Create a calculate_percentage() method to calculate the percentage.
6. Create a check_result() method:
o Student passes if marks in every subject are 35 or above.
o Otherwise, display FAIL.
7. Create an update_marks() method to update the marks of a selected subject.

```python
class Student:
    def __init__(self,stud_name,stud_rollno,stud_age,sub1mark,sub2mark,sub3mark):
        self.stud_name = stud_name
        self.stud_rollno = stud_rollno
        self.stud_age = stud_age
        self.sub1mark = sub1mark
        self.sub2mark = sub2mark
        self.sub3mark = sub3mark

    def display_detail(self):
        print("name:-",self.stud_name) 
        print("roll no",self.stud_rollno)
        print("age",self.stud_age)
        print("sub1marks",self.sub1mark) 
        print("sub2marks",self.sub2mark) 
        print("sub3marks",self.sub3mark)
        print("") 

    def calculate_total(self):
        total = self.sub1mark + self.sub2mark + self.sub3mark
        return total
        print("")

    def calculate_percentage(self):
        total = self.calculate_total()
        percentage = total / 3
        return percentage

    def check_result(self):
        if self.sub1mark>=35 and self.sub2mark>=35 and self.sub3mark>=35:
            print("Pass")
        else:
            print("Fail")
  
    def update_mark(self,subject,new_mark):
        if subject == 1:
            self.sub1mark= new_mark
        elif subject == 2:
            self.sub2mark= new_mark
        elif subject == 3:
            self.sub3mark= new_mark
        else:
            print("invalid subject")
            return
        print("mark update succesfully")

std1=Student("dipali",29,21,35,86,71)
std1.display_detail()

print("total marks",std1.calculate_total())

print("calculate percentage",std1.calculate_percentage(),"%")

print("check result",std1.check_result())

print("update mark",std1.update_mark(2,90))

print("update info",std1.display_detail())

print(" new total marks",std1.calculate_total())

print("calculate new percentage",std1.calculate_percentage(),"%")

print("check result",std1.check_result())











         
```

2. Create a Library Management System using Class and Object in Python.
What to Do
1. Create a class named Book.
2. Create a constructor __init__() to initialize:
o Book name
o Book ID
o Author name
o Availability status
3. Create a display_book() method to display book details.
4. Create a issue_book() method:
o Check whether the book is available.
o If available, issue the book and change its status.
o If already issued, display an appropriate message.
5. Create a return_book() method:
o Return the issued book.
o Change its availability status back to available.
6. Create a check_availability() method to display whether the book is available or
issued.


```python
class Book:
    def __init__(self,Book_name,Book_id,author_name,availability_status):
        self.Book_name = Book_name
        self.Book_id = Book_id
        self.author_name = author_name
        self.availability = availability_status

    def display_book(self):
        print("book name:-",self.Book_name)
        print("book id:-",self.Book_id)
        print("author name:-",self.author_name)
        print("book availability:-",self.availability)

    def issue_book(self):
        if self.availability == True:
            self.availability = False
            print("Book issued successfully")
        else:
            print("Book is already issued")

    def return_book(self):
        if self.availability == False:
            self.availability = True
            print("Book returned successfully")
        else:
            print("Book is already available")

    def check_availability(self):
        if self.availability == True:
            print("Book is available")
        else:
            print("Book is issued")
            
book1 = Book("Python Programming", 101, "John Smith", False)
book1.check_availability()
book1.issue_book()
book1.return_book()





        
```

3. Create a simple Employee Management System using Class and Object in Python.
What to Do
1. Create a class named Employee.
2. Create a constructor __init__() to initialize:
o Employee name
o Employee ID
o Department
o Basic salary
3. Create a display_details() method to display employee information.
4. Create a calculate_salary() method:
o Add a fixed bonus of ₹5,000.
o Calculate and display the final salary.
5. Create a check_salary() method:
o If salary is ₹30,000 or above, display "Good Salary".
o Otherwise, display "Average Salary".
6. Create a menu-driven program:
o 1 → Display Details
o 2 → Calculate Salary
o 3 → Check Salary
o 4 → Exit

```python
class Employee:

    def __init__(self, name, employee_id, department, basic_salary):
        self.name = name
        self.employee_id = employee_id
        self.department = department
        self.basic_salary = basic_salary

    def display_details(self):
        print("\n--- Employee Details ---")
        print("Employee Name:", self.name)
        print("Employee ID:", self.employee_id)
        print("Department:", self.department)
        print("Basic Salary: ₹", self.basic_salary)

    def calculate_salary(self):
        bonus = 5000
        final_salary = self.basic_salary + bonus

        print("\n--- Salary Calculation ---")
        print("Basic Salary: ₹", self.basic_salary)
        print("Bonus: ₹", bonus)
        print("Final Salary: ₹", final_salary)

    def check_salary(self):
        if self.basic_salary >= 30000:
            print("Good Salary")
        else:
            print("Average Salary")

# Creating Employee object
emp1 = Employee("Dipali", 101, "IT", 35000)

# Menu-driven program
while True:
    print("\n===== Employee Management System =====")
    print("1. Display Details")
    print("2. Calculate Salary")
    print("3. Check Salary")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        emp1.display_details()

    elif choice == 2:
        emp1.calculate_salary()

    elif choice == 3:
        emp1.check_salary()

    elif choice == 4:
        print("Thank you!")
        break

    else:
        print("Invalid choice!")
```
