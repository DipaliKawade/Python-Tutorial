```python
#find mid
list1 = [1, 2, 3]
list2 = [7, 4, 5, 6, 8]

res = list1 + list2

mid = len(res) // 2

if len(res) % 2 == 0:
    # Even: two middle elements
    med = (res[mid - 1] + res[mid]) / 2
else:
    # Odd: one middle element
    med = res[mid]

print(med)

print("")
#another example

list1 = [1, 2, 3]
list2 = [7, 4, 5, 6, 8]

res = list1 + list2

if len(res) % 2 == 1:
    # Odd: one middle element
    mid = len(res) // 2
    print(res[mid])

else:
    # Even: two middle elements
    mid = len(res) // 2
    front = mid - 1

    med = (res[mid] + res[front]) / 2
    print(med)

print("")
#another one

list1 = [1.1, 2.2, 3.6]
list2 = [7.2, 4.1, 5, 6, 8]

res = list1 + list2

if len(res) % 2 == 1:           # Odd: one middle element
    mid = len(res) // 2
    print(res[mid])

else:                           # Even: two middle elements
    mid = len(res) // 2
    front = mid - 1

    med = (res[mid] + res[front]) / 2
    print(med)
```
