# a magic method also called as Dunder Method it is a special method in python whose name start and ends with __name__
# it is automatically called when an operation performs on object

```python
class demo:
    def __init__(self,num1):
        self.num1 = num1

    def __str__(self):              #__str__ method
        return self.num1
    def __len__(self):
        return len(self.num1)
    def __eq__(val1, val2):
        return val1.num1 == val2.num1

obj1=demo("10")
print(obj1)

print()

obj2=demo("demo")
print(len(obj2))
print(obj2)

print()

print(obj1 == obj2)
```

```python

class Temp:
    def __init__(self,num1):
        self.num1 = num1

    def __add__(val1,val2):
        return val1.num1 + val2.num1
    
    def __sub__(val1,val2):
        return val1.num1 - val2.num1
    
    def __truediv__(val1,val2):
        return val1.num1 / val2.num1

    def __mul__(val1,val2):
        return val1.num1 * val2.num1
    
    def __floordiv__(val1,val2):
        return val1.num1 // val2.num1
    
    def __mod__(val1,val2):
        return val1.num1 % val2.num1
        


obj1=Temp(20)
obj2=Temp(10)

print(obj1 + obj2)
print(obj1 - obj2)
print(obj1 / obj2)
print(obj1 * obj2)
print(obj1 // obj2)
print(obj1 % obj2)


```
