```python
#zeroDivisionError
try:
    num1 = 10
    num2 = 0

    result = num1 / num2   # This will raise ZeroDivisionError

except ZeroDivisionError as e:
    print("An error occurred:", e)

else:
    print("Result:", result)
```

```python
# valueError
# ValueError

try:
    num = int(input("Enter a number: "))   # Enter abc

except ValueError as e:
    print("An error occurred:", e)

else:
    print("You entered:", num)
```

```python
# TypeError

try:
    num = 10
    text = "20"

    result = num + text    # This will raise TypeError

except TypeError as e:
    print("An error occurred:", e)

else:
    print("Result:", result)
```

```python
#indexError
# list1 = [1, 2, 3, 4, 5]
# print("List:", list1)
# print(list1[10])  # This will raise an IndexError since index 10 is out of range


list1 = [1, 2, 3, 4, 5]
try:
    temp = list1[10]  # This will raise an IndexError since index 10 is out of range
except IndexError as e:
    print("An error occurred:", e)
else:
    print("Accessed element:", temp)
```

```python
# keyError

# student = {
#     "name": "Dipali",
#     "age": 22
# }

# print(student["marks"])    # This will raise KeyError

# Print("fix this error")

# KeyError Handling using try-except-else

student = {
    "id": 101,
    "name": "Dipali",
    "marks": 95
}

try:
    age = student["age"]   # Key 'age' does not exist

except KeyError as e:
    print("Key not found:", e)

else:
    print("Age:", age)
```

```python
# FileNotFoundError

# file = open("student.txt", "r")
# print(file.read())

# file.close()

# print("fix the error")

# FileNotFoundError Handling

try:
    file = open("student.txt", "r")

except FileNotFoundError:
    print("File not found.")

else:
    print(file.read())
    file.close()
```
