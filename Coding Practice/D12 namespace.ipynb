```python
""" 4. What will be the output? Is x being shadowed? """
x = 100

def outer():
    x = 200

    def inner():
        x = 300
        print(x)

    inner()
    print(x)

outer()
print(x)

"""Yes, x is being shadowed.

There are three different x variables:

Global x     = 100
    ↓
outer x     = 200
    ↓
inner x     = 300

Inside inner():

print(x)

uses local x = 300.

Inside outer():

print(x)

uses outer's x = 200.

Outside everything:

print(x)

uses global x = 100.

Shadowing means an inner variable has the same name as a variable in an outer namespace."""
```

```python
""" 5. What will happen when this program runs? Explain why"""
count = 10

def update():
    count = count + 1
    print(count)

update()
"""This gives:

UnboundLocalError

Why?

Python sees:

count = count + 1

and because count is assigned inside update(), Python considers count a Local variable.

So Python effectively tries:

Local count = Local count + 1

But the local count has not been given a value yet.

Therefore:

UnboundLocalError: local variable 'count' referenced before assignment
Python does not automatically use the global count once 
it has determined that count is local to the function."""
```

```python
"""6. Correct the following program using the global keyword so that the final output is 20."""
x = 10

def update():
    global x
    x = 20

update()
print(x)

"""The problem is that x = 20 creates a local x.

Use global:

x = 10

def update():
    global x
    x = 20

update()
print(x)

Output:

20

global x tells Python:

"Don't create a local x. Use the x from the Global namespace."

So the global value changes:

Before update() → x = 10
After update()  → x = 20"""
```

```python
""" 7. What will be the output? Explain the difference between the Local and Enclosing
Namespace. """
def outer():
    x = 10

    def inner():
        x = 20
        print(x)

    inner()
    print(x)

outer()

"""When inner() executes:

print(x)

Python first searches the Local namespace of inner() and finds 20.

After inner() finishes:

print(x)

inside outer() finds outer's local x = 10."""
```

```python
""" 8. What will be the output? Why does nonlocal change the value of x? """
def outer():
    x = 10

    def inner():
        nonlocal x
        x = 50

    inner()
    print(x)

outer()

""" Without nonlocal:

def inner():
    x = 50

Python would create a new local x inside inner().

But:

nonlocal x

means:

"Use the x from the nearest enclosing function."

So inner() changes:

outer x: 10 → 50
global vs   nonlocal
Keyword	    Changes
global	    Global variable
nonlocal	Enclosing function's variable """
```
