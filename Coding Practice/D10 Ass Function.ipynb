```python
""" 1. Write a program using a lambda function to find the square of each number in a list
using map().  """

list1 = [1, 2, 3, 4, 5]

result = list(map(lambda x: x * x, list1))

print(result)
```

```python
"""" 2. Write a program using a lambda function to convert all names in a list to uppercase
using map().  """ 

list1 = ["dipali","sahil","divya","siddhu"]

result = list(map(lambda x: x.upper(),list1))

print(result)
```

```python
"""" 3. Write a program using map() to calculate a 10% discount on a list of product prices. """

list1 = [1, 2, 3, 4, 5]

result = list(map(lambda x: x - x * 0.10,list1))

print(result)
```

```python
""" 4. Write a program using filter() to extract all even numbers from a given list. """

list1 = [1, 2, 3, 4, 5]

result = list(filter(lambda x: x % 2==0,list1))

print(result)
```

```python
"""5. Write a program using filter() to display all students who scored 35 or more marks. """
marks_list = [25, 35, 40, 18, 60, 33]

passed_list = list(filter(lambda x: x >= 35, marks_list))

print(passed_list)
```

```python
""" 6. Write a program using filter() to remove all empty strings from a list.  """

list1 = ["dipali","","sahil","divya","","siddhu"]

result = list(filter(lambda x: x!="", list1))

print(result)
```

```python
"""7. Write a program using map() to find the length of each string in a list. """

list1=["dipali","sahil","divya","siddhu"]

result=list(map(lambda x:len(x),list1))

print(result)
```

```python
""" 8. Write a program using reduce() to calculate the sum of all numbers in a list.  """

print("using reduce function")
from functools import reduce

listOne=[1,2,3,4,5]  # Reduces the list to a single value by adding all elements

print(listOne)

result=reduce(lambda x, y: x + y, listOne)
print(result)

#another way
print()
print("using sum function:-")
listOne=[1,2,3,4,5] 

print(sum(listOne))
```

```python
""" 9. Given a list of employee salaries, use map() to increase each salary by 15%.  """

list1 = [10000, 20000, 30000, 40000, 50000]

result = list(map(lambda x: x + x * 0.15,list1))

print(result)
```

```python
""" 10. Given a list of integers, use filter() to display only the numbers that are divisible by
both 3 and 5.  """

list1 = [15, 20, 30, 21, 45,28]

result = list(filter(lambda x: x % 3 == 0 and x % 5 == 0,list1))

print(result)
```
