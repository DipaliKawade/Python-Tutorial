```python
# Q-1 Write a program to count the number of vowels and consonants in a given string.
string = input("Enter a string: ")

vowels = 0
consonants = 0

for i in string:
    if i.isalpha():      # Check if it is a letter
        if i in "aeiouAEIOU":
            vowels += 1
        else:
            consonants += 1

print("Vowels =", vowels)
print("Consonants =", consonants)
```

```python
# Q-2 Write a program to check whether a given string is a Palindrome.
string = input("Enter a string: ")

reverse = ""

for i in string:
    reverse = i + reverse

if string == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# Q-3 Write a program to reverse a string without using slicing ([::-1]) or built-in functions.
string = input("Enter a string: ")

reverse = ""

for i in string:
    reverse = i + reverse

print("Reversed String =", reverse)
```

```python
# Q-4 Write a program to count the occurrence of each character in a string
string = input("Enter a string: ")

for i in string:
    print(i, ":", string.count(i))
```

```python
"""Q-5 Write a program to remove all duplicate characters from a string while maintaining their
original order"""

string = input("Enter a string: ")

newWord = ""

for i in string:
    if i not in newWord:
        newWord = newWord + i

print(newWord)
```

```python
# Q-6 Write a program to find the longest word in a sentence.
string = input("Enter a sentence: ")

word = ""
longest = ""

for i in string:
    if i != " ":
        word = word + i
    else:
        if len(word) > len(longest):
            longest = word
        word = ""

if len(word) > len(longest):
    longest = word

print("Longest Word =", longest)
```

```python
# Q-7 Write a program to capitalize the first letter of every word without using the title()
method
str1= input("Enter a String")
word = str1.split()
for i in word:
    print(i.capitalize(),end=" ")
```

```python
# Q-8 Count Total Words (Without split())
string = input("Enter a sentence: ")

count = 1

for i in string:
    if i == " ":
        count = count + 1

print("Total Words =", count)
```

```python
# Q-9 Remove All Spaces in string
str1 = input("Enter a string: ")
sen = ""
for i in str1:
    if i != " ":
        sen = sen + i
print(sen)
```
