```python
# 1. Write a function that accepts two numbers and returns the sum, difference, using
# multiple return values. 

def calculate_sum_and_difference(num1, num2):
    sum_result = num1 + num2
    difference_result = num1 - num2
    return sum_result, difference_result
print(calculate_sum_and_difference(10, 5)) 
```

```python
# 2. Function to return Largest and Smallest Number
def find_largest_and_smallest(num1, num2, num3):
    largest = max(num1, num2, num3)
    smallest = min(num1, num2, num3)
    return largest, smallest

print(find_largest_and_smallest(10, 5, 15))
print("largest number is", find_largest_and_smallest(10, 5, 15)[0])
print("smallest number is", find_largest_and_smallest(10, 5, 15)[1])

print()

#another way to do it is to unpack the values into two variables using(*args) it is used to pass a variable number of arguments to a function. It allows you to pass any number of positional arguments to the function, and they will be collected into a tuple. This is useful when you want to create functions that can accept a flexible number of inputs.

def find_largest_and_smallest(*args):
    largest = max(args)
    smallest = min(args)
    return largest, smallest
large,small=find_largest_and_smallest(10, 5, 15, 20, 3)
print("largest number is", large)
print("smallest number is", small)


```

```python
"""3. Write a function using *args to calculate the sum, average, maximum, and minimum of any number of integers. """

def calculate(*args):
    total = sum(args)
    average = total / len(args)
    maximum = max(args)
    minimum = min(args)

    return total, average, maximum, minimum

sum, avg, max, min = calculate(10, 20, 30, 40, 50)

print("Sum =", sum)
print("Average =", avg)
print("Maximum =", max)
print("Minimum =", min)
```

```python
"""4. Write a function using **kwargs to display employee details (id, name, department,
salary) in a formatted manner"""

def display_employee_details(**kwargs):
    print("Employee Details:")

    for key, value in kwargs.items():
        print(f"{key}: {value}")
        
display_employee_details(id=101, name="Dipali", department="IT", salary=50000)
```

```python
""" 5. Write a function that accepts a string and returns the number of uppercase letters,
lowercase letters, digits. """

def count_string_components(input_string):
    uppercase_count = 0
    lowercase_count = 0
    digit_count = 0

    for char in input_string:
        if char.isupper():
            uppercase_count += 1
        elif char.islower():
            lowercase_count += 1
        elif char.isdigit():
            digit_count += 1

    return uppercase_count, lowercase_count, digit_count

text = input("Enter a string: ")
print("You entered:", text) 
upper, lower, digits = count_string_components(text)

print("Uppercase letters:", upper)
print("Lowercase letters:", lower)
print("Digits:", digits)
```

```python
# 6. Write a function that checks whether a given string is a Palindrome. 

def palindrome(text):
    if text == text[::-1]:
        return "Palindrome"
    else:
        return "Not Palindrome"

word = input("Enter a string: ")
print("You entered:", word)
print(palindrome(word))
```

```python
"""7. Write a function that accepts a list of integers and returns a new list containing only
even numbers. """

def even_numbers(list1):
    even = []

    for i in list1:
        if i % 2 == 0:
            even.append(i)

    return even

numbers = [10, 15, 20, 33, 42, 55, 60]

result = even_numbers(numbers)

print("Original List:", numbers)
print("Even Numbers:", result)
```

```python
""" 8. Create a menu-driven calculator using functions that performs Addition,
Subtraction, Multiplication, Division, and Exit. """

def add(a, b):
    return a + b

def sub(a, b):
    return a - b

def mul(a, b):
    return a * b

def div(a, b):
    return a / b

while True:

    print("\n----- Calculator -----")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 5:
        print("Your Choice =", choice)
        print("Program Ended")
        break

    num1 = int(input("Enter First Number: "))
    num2 = int(input("Enter Second Number: "))
      # Display the entered values
    print("Your Choice =", choice)
    print("First Number =", num1)
    print("Second Number =", num2)

    match choice:

        case 1:
            print("Addition =", add(num1, num2))

        case 2:
            print("Subtraction =", sub(num1, num2))

        case 3:
            print("Multiplication =", mul(num1, num2))

        case 4:
            if num2 == 0:
                print("Cannot divide by zero")
            else:
                print("Division =", div(num1, num2))

        case _:
            print("Invalid Choice")
```

```python
""" 9. Write a function that accepts a dictionary of students and their marks, then returns
the name of the topper and the highest marks. """
print("1st logic")
def topper(student):

    topper_name = max(student, key=student.get)
    highest_marks = student[topper_name]

    return topper_name, highest_marks

students = { "Dipali": 85,"Sahil": 98,"Divya": 88,"Shubh": 95}

name, marks = topper(students)

print("Topper Name:", name)
print("Highest Marks:", marks)

# logic 2
print()
print("2nd logic")

def topper(student):

    topper_name = ""
    highest_marks = 0

    for key, value in student.items():

        if value > highest_marks:
            highest_marks = value
            topper_name = key

    return topper_name, highest_marks

students = { "Dipali": 85,"Sahil": 98,"Divya": 88,"Shubh": 95}

topper_name, highest_marks = topper(students)

print("Topper Name:", topper_name)
print("Highest Marks:", highest_marks)
```

```python
""" 10. Write a function that accepts a sentence and returns the frequency of each word
using a dictionary.  """

def word_frequency(sentence):

    words = sentence.split()
    freq = {}

    for word in words:
        freq[word] = words.count(word)

    return freq

text = input("Enter a sentence: ")

print(word_frequency(text))
```

```python
""" 11. Write a function that accepts a list of integers and returns:
• Sum of all numbers
• Count of even numbers
• Count of odd numbers  """

def list1(numbers):

    total = sum(numbers)

    even = list(filter(lambda x: x % 2 == 0, numbers))
    odd = list(filter(lambda x: x % 2 != 0, numbers))

    return total, len(even), len(odd)

num1 = [10, 15, 20, 25, 30, 35]

sum1, even_count, odd_count = list1(num1)

print("Sum =", sum1)
print("Even Count =", even_count)
print("Odd Count =", odd_count)
```
