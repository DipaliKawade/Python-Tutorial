```python
print("----- Primitive Data Types -----")

# Integer
integer_value = 10
print("Value =", integer_value)
print("Data Type =", type(integer_value))
print("Memory Address =", id(integer_value))

# Float
float_value = 10.5
print("\nValue =", float_value)
print("Data Type =", type(float_value))
print("Memory Address =", id(float_value))

# Complex
complex_value = 2 + 3j
print("\nValue =", complex_value)
print("Data Type =", type(complex_value))
print("Memory Address =", id(complex_value))

# Boolean
boolean_value = True
print("\nValue =", boolean_value)
print("Data Type =", type(boolean_value))
print("Memory Address =", id(boolean_value))

# String
string_value = "Python"
print("\nValue =", string_value)
print("Data Type =", type(string_value))
print("Memory Address =", id(string_value))
```

```python
print("----- Arithmetic Operations -----")

a = 15
b = 4

print("Addition =", a + b)
print("Subtraction =", a - b)
print("Multiplication =", a * b)
print("Division =", a / b)
print("Floor Division =", a // b)
print("Modulus =", a % b)
print("Exponent =", a ** b)
```

```python
print("----- Type Conversion -----")

x = "100"
y = 50

# Convert string to integer
x = int(x)

print("Addition =", x + y)
print("Subtraction =", x - y)
print("Multiplication =", x * y)
print("Division =", x / y)
```

```python
print("----- Electricity Bill Calculation -----")

# Variables
units = 250
cost_per_unit = 8

# Calculate Bill
total_bill = units * cost_per_unit
gst = total_bill * 0.18
final_bill = total_bill + gst

print("Units Consumed =", units)
print("Cost per Unit =", cost_per_unit)
print("Total Bill =", total_bill)
print("GST (18%) =", gst)
print("Final Bill =", final_bill)
```

```python
print("----- Membership Operators -----")

languages = ["Python", "C", "C++", "JavaScript"]

print("Python exists :", "Python" in languages)
print("Java does not exist :", "Java" not in languages)
print("C++ exists :", "C++" in languages)
```

```python
print("----- Type Conversion -----")

# int → float
a = 10
b = float(a)
print("int to float :", b)

# float → string
c = 15.5
d = str(c)
print("float to string :", d)

# string → int
e = "100"
f = int(e)
print("string to int :", f)

# int → bool
g = 1
h = bool(g)
print("int to bool :", h)

# bool → int
i = True
j = int(i)
print("bool to int :", j)
```

```python
print("----- Product Information -----")

# Product Details
product_name = "Laptop"
price = 50000
quantity = 2

# Calculations
total_cost = price * quantity
gst = total_cost * 0.18
final_amount = total_cost + gst

# Display
print("Product Name :", product_name)
print("Price :", price)
print("Quantity :", quantity)
print("Total Cost :", total_cost)
print("GST (18%) :", gst)
print("Final Amount :", final_amount)
```
