```python
# Q1 Write a program to find the union of two sets without using the union() method

setOne = {10, 20, 30}
setTwo = {30, 40, 50}

newSet = setOne.copy()

for i in setTwo:
    newSet.add(i)

print(newSet)
```

```python
# Q2 Write a program to remove all duplicate elements from a list using a set

listOne = [10, 20, 10, 30, 20, 40]

newSet = set(listOne)

print(newSet)
```

```python
# Q3 Write a program to find all common elements between two lists using sets.

listOne = [10, 20, 30, 40]
listTwo = [20, 30, 50]

setOne = set(listOne)
setTwo = set(listTwo)

print(setOne & setTwo)
```

```python
# Q4 Write a program to count the number of unique elements in a list using a set.
listOne = [10, 20, 10, 30, 20, 40]

newSet = set(listOne)

print("Unique Elements =", len(newSet))
```

```python
# Q5 Write a program to find the missing numbers between 1 to N using sets.
N = 10

numbers = {1, 2, 3, 5, 7, 8, 10}

allNumbers = set(range(1, N + 1))

print(allNumbers - numbers)
```

```python
"""Q6 Write a program to create a new set containing only the even numbers from a given
set"""
setOne = {10, 15, 20, 25, 30}

even = set()

for i in setOne:
    if i % 2 == 0:
        even.add(i)

print(even)
```

```python
""" Q7 Write a program to merge three sets into a single set containing only unique
elements"""

setOne = {10, 20}
setTwo = {20, 30}
setThree = {30, 40}

newSet = setOne.copy()

for i in setTwo:
    newSet.add(i)

for i in setThree:
    newSet.add(i)

print(newSet)
```

```python
# Q8 Given two strings, find the common characters using sets.

strOne = "dipali"
strTwo = "kawade"

setOne = set(strOne)
setTwo = set(strTwo)

print(setOne & setTwo)
```

```python
# Q9 Given multiple lists, find all elements that appear in every list using sets.

listOne = [1, 2, 3, 4]
listTwo = [2, 3, 5]
listThree = [2, 3, 6]

setOne = set(listOne)
setTwo = set(listTwo)
setThree = set(listThree)

print(setOne & setTwo & setThree)
```
