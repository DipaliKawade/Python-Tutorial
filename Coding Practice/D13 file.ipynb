```python
"""1. Write a program to create a text file and store the name, age, and marks of a student
in it. """

name = input("Enter student name: ")
age = int(input("Enter age: "))
marks = float(input("Enter marks: "))

with open("student.txt", "w") as file:
    file.write("Name: " + name + "\n")
    file.write("Age: " + str(age) + "\n")
    file.write("Marks: " + str(marks) + "\n")

print("Student record stored successfully.")
```

```python
"""2. Write a program to read the complete contents of a file and count the total number
of characters. """

with open("student.txt", "r") as file:
    data = file.read()

print("File contents:")
print(data)

print("Total characters:", len(data))
```

```python
"""3. Write a program to read a file line by line and count the total number of lines. """

count = 0

with open("student.txt", "r") as file:
    for line in file:
        count += 1

print("Total number of lines:", count)
```

```python
"""4. Write a program to read a file and count the number of words present in it. """

with open("student.txt", "r") as file:
    data = file.read()

words = data.split()

print("Total number of words:", len(words))
```

```python
"""5. Write a program to copy the contents of one text file into another file. """

with open("source.txt", "r") as source:
    data = source.read()

with open("destination.txt", "w") as destination:
    destination.write(data)

print("File copied successfully.")
```

```python
"""6. Write a program to read a file and create another file containing only the lines that
contain the word "Python". """

with open("source.txt", "r") as source:
    with open("python_lines.txt", "w") as destination:
        for line in source:
            if "Python" in line:
                destination.write(line)

print("Python lines copied successfully.")
```

```python
"""7. Write a program to append new student records to an existing file without deleting
the previous records. """

name = input("Enter student name: ")
age = int(input("Enter age: "))
marks = float(input("Enter marks: "))

with open("student.txt", "a") as file:
    file.write("Name: " + name + "\n")
    file.write("Age: " + str(age) + "\n")
    file.write("Marks: " + str(marks) + "\n")
    file.write("\n")

print("New student record added successfully.")
```

```python
"""8. Write a program to read a file and find the longest word present in it. """

with open("source.txt", "r") as file:
    data = file.read()

words = data.split()

longest = max(words, key=len)

print("Longest word:", longest)
print("Length:", len(longest))
```

```python
"""9. Write a program to read a file containing numbers and calculate their sum, average,
maximum, and minimum."""

with open("numbers.txt", "r") as file:
    data = file.read()

numbers = data.split()

numbers = [int(num) for num in numbers]

total = sum(numbers)
average = total / len(numbers)
maximum = max(numbers)
minimum = min(numbers)

print("Sum:", total)
print("Average:", average)
print("Maximum:", maximum)
print("Minimum:", minimum)
```
