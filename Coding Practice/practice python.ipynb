```python
#print the following pattern
row = 4

# Increasing Pattern
space = row - 1
star = 1

for i in range(row):
    print(" " * space + "*" * star)
    space -= 1
    star += 1

# Decreasing Pattern
space = 1
star = row - 1

for i in range(row - 1):
    print(" " * space + "*" * star)
    space += 1
    star -= 1

    
# using another way
print()
row = 4

# Increasing
for i in range(1, row + 1):
    print("*" * i)

# Decreasing
for i in range(row - 1, 0, -1):
    print("*" * i)



print()
row = 4

# Increasing
for i in range(1, row + 1):
    print(" " * (row - i) + "*" * i)

# Decreasing
for i in range(row - 1, 0, -1):
    print(" " * (row - i) + "*" * i)
```

```python
def separate_numbers(lst):
    positive = []
    negative = []
    zero = []

    for i in lst:
        if i > 0:
            positive.append(i)
        elif i < 0:
            negative.append(i)
        else:
            zero.append(i)

    return positive, negative, zero


# Example
nums = [10, -5, 0, 20, -8, 15, 0, -2]

p, n, z = separate_numbers(nums)

print("Positive:", p)
print("Negative:", n)
print("Zero:", z)
```

```python
def highest_marks(students):
    topper = max(students, key=students.get)
    print("Topper:", topper)
    print("Marks:", students[topper])


# Example
students = {
    "dipali": 85,
    "Priya": 92,
    "sahil": 78,
    "seeta": 95
}

highest_marks(students)
```
