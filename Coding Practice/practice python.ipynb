```python
def separate_numbers(lst):
    positive = []
    negative = []
    zero = []

    for i in lst:
        if i > 0:
            positive.append(i)
        elif i < 0:
            negative.append(i)
        else:
            zero.append(i)

    return positive, negative, zero


# Example
nums = [10, -5, 0, 20, -8, 15, 0, -2]

p, n, z = separate_numbers(nums)

print("Positive:", p)
print("Negative:", n)
print("Zero:", z)
```

```python
def highest_marks(students):
    topper = max(students, key=students.get)
    print("Topper:", topper)
    print("Marks:", students[topper])


# Example
students = {
    "dipali": 85,
    "Priya": 92,
    "sahil": 78,
    "seeta": 95
}

highest_marks(students)
```
