```python
s = "Python"

# 1. Ordered
print("Ordered:", s)

# 2. Indexed
print("Index:", s[0])

# 3. Slicing
print("Slicing:", s[1:4])

# 4. Iterable
print("Iterable:")
for ch in s:
    print(ch)

# 5. Allows duplicates
word = "banana"
print("Duplicates:", word)

# 6. Homogeneous
print("Homogeneous:", "ABC123")

# 7. Positive indexing
print("Positive index:", s[2])

# 8. Negative indexing
print("Negative index:", s[-1])

# 9. Concatenation
print("Concatenation:", "Hello" + " World")

# 10. Repetition
print("Repetition:", "Hi " * 3)

# 11. Sequence
print("Sequence:", s)

# 12. Dynamic memory
s = "Hello Python Programming"
print("Dynamic:", s)
```

```python
str1 = " hello world "

print(str1.upper())
print(str1.lower())
print(str1.capitalize())
print(str1.title())
print(str1.swapcase())

print(str1.strip())
print(str1.lstrip())
print(str1.rstrip())

print(str1.replace("world", "python"))

print(str1.split())

print("-".join(["hello", "world"]))

print(str1.find("world"))
print(str1.index("world"))
print(str1.count("o"))

print(str1.startswith(" hello"))
print(str1.endswith(" "))

print("hello".isalpha())
print("123".isdigit())
print("hello123".isalnum())
```

```python
name = "hello"

print("String:", name)
print("Length:", len(name))
print("Maximum:", max(name))
print("Minimum:", min(name))
print("Sorted:", sorted(name))
print("Type:", type(name))

print("List:", list(name))
print("Tuple:", tuple(name))
print("Set:", set(name))

print("String:", str(123))

print("Enumerate:", list(enumerate(name)))

print("Reverse:", ''.join(reversed(name)))

print("ASCII of A:", ord("A"))
print("Character:", chr(65))

print("Any:", any([False, True, False]))
print("All:", all([True, True, True]))
```
