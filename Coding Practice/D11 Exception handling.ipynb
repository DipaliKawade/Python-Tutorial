```python
# Handle ZeroDivisionError
try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    result = num1 / num2
    print("Result =", result)

except ZeroDivisionError as e:
    print("Cannot divide by zero",e)
```

```python
# Handle ValueError
try:
    age = int(input("Enter your age: "))
    print("Your age is", age)

except ValueError as e:
    print("Please enter numbers only.",e)
```

```python
# Handle IndexError
list1 = [10, 20, 30]

try:
    index = int(input("Enter index: "))
    print(list1[index])

except IndexError as e:
    print("your index is:-",index)
    print("Index is out of range.",e)
```

```python
# Handle KeyError
student = {
    "name": "Dipali",
    "age": 21
}

try:
    key = input("Enter key: ")
    print(student[key])

except KeyError as e:
    print("Your enter key:-",key)
    print("Key not found.",e)
```

```python
# Handle TypeError
try:
    num = 10
    text = "20"

    print(num + text)

except TypeError as e:
    print("Cannot add integer and string.",e)
```

```python
# Use try, except, else, and finally
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    result = a / b

except ZeroDivisionError as e:
    print("Cannot divide by zero.",e)

else:
    print("Result =", result)

finally:
    print("Program finished.")
```
