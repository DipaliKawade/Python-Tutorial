```python
print("position arguments:-")
def info(name,rollno):
    print("student name is",name,"student roll no is",rollno)
info("Dipali",10)

print()
print("without position arguments:-")
def info(name,rollno):
    print("student name is",name,"student roll no is",rollno)
info(10,"Dipali")

print()
print("keyword arguments:-")
def info(name,rollno):
    print("student name is",name,"student roll no is",rollno)
info(rollno=10,name="Dipali")

print()
print("default arguments:-")
def info(name="dipali",rollno=10):
    print("student name is",name,"student roll no is",rollno)
info()
info("sidddhu",15)

print()
print("variable length arguments:-")
# def info(*args):
#     print("student name is",args[0],"student roll no is",args[1])
# info("Dipali",10) 


def temp(*args):
    for i in args:
        print(i)
temp(10,20,30,40,50)


print()
print("variable length keyword arguments:-")
def info(**kwargs):         # kwargs is a dictionary
    print(kwargs)
info(name="Dipali",rollno=10)

print()
print("ex2:-")
def detail(**kwargs):
    for key,value in kwargs.items():
        print(key,value)
detail(name="Dipali",rollno=10,age=20,city="pune")



```

```python
print()
print("nested function:-")
def outer():
    print("this is outer function")
    def inner():
        print("this is inner function")
    inner()
outer()


print()
print("lenn")
def lenn(arg):
    count=0
    for i in arg:
        count+=1
    return count
print(lenn("Dipali"))


print()
print("global variable:-")
x=10
print(x)

print()
print("local variable:-")
def temp():
    x=20
    print(x)
temp()

# lambda function:- lambda function is a small anonymous function. It can take any number of arguments, but can only have one expression.
print()
print("lambda function:-")
def square(x):
    return x * x
print(square(5))

print()
print("lambda function:-")
square = lambda x: x * x
print(square(5))

```
