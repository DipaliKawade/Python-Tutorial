```python
""" Q1. Write a program to count the frequency of each character in a string using a
dictionary"""

string = "programming"

count = {}

for i in string:
    if i in count:
        count[i] += 1
    else:
        count[i] = 1

print(count)
```

```python
# Q2 Write a program to count the frequency of each element in a list using a dictionary.

listOne = [10, 20, 10, 30, 20, 10]

count = {}

for i in listOne:
    if i in count:
        count[i] += 1
    else:
        count[i] = 1

print(count)
```

```python
# Q3 Write a program to find the key having the maximum value in a dictionary
student = {'A': 85, 'B': 92, 'C': 78}

key = ""
maxValue = 0

for i in student:
    if student[i] > maxValue:
        maxValue = student[i]
        key = i

print("Key =", key)
print("Value =", maxValue)
```

```python
# Q4 Write a program to find the key having the minimum value in a dictionary.

student = {'A':85, 'B':92, 'C':78}

key = ""
minValue = 100

for i in student:
    if student[i] < minValue:
        minValue = student[i]
        key = i

print(key)
```

```python
# Q5 Write a program to merge two dictionaries into a single dictionary.
dict1 = {'a':1, 'b':2}
dict2 = {'c':3, 'd':4}

dict1.update(dict2)

print(dict1)
```

```python
# Q6 Write a program to remove duplicate values from a dictionary
dictOne = {'a':10, 'b':20, 'c':10, 'd':30}

newDict = {}

for i in dictOne:
    if dictOne[i] not in newDict.values():
        newDict[i] = dictOne[i]

print(newDict)
```

```python
# Q7 Write a program to check whether a given key exists in a dictionary.

student = {'A':85, 'B':92, 'C':78}

key = input("Enter Key: ")

if key in student:
    print("Key Found")
else:
    print("Key Not Found")
```

```python
# Q8 Write a program to find the sum of all values in a dictionary.

marks = {'Math':85, 'Science':90, 'English':80}

total = 0

for i in marks:
    total += marks[i]

print(total)
```

```python
"""Q9 Write a program to create a dictionary from two lists: one containing keys and the
other containing values."""

keys = ['A', 'B', 'C']
values = [10, 20, 30]

student = {}

for i in range(len(keys)):
    student[keys[i]] = values[i]

print(student)
```

```python
""" Q10 Write a program to create a student result dictionary containing name and marks of
5 students, then display the topper"""

student = {
    'Ram':85,
    'Sham':90,
    'Amit':78,
    'Neha':95,
    'Riya':88
}

topper = ""
maxMarks = 0

for i in student:
    if student[i] > maxMarks:
        maxMarks = student[i]
        topper = i

print("Topper =", topper)
print("Marks =", maxMarks)
```
