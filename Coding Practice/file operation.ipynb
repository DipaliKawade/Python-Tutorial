```python
str1 = " hello "
str2= " hello user1 "
fileobj=open("temp.txt","w")
fileobj.write(str2)
fileobj.close()

str1 = " hello "
str2= " hello user1 "
fileobj=open("temp.txt","a")
fileobj.write(str1)
fileobj.close()
```

```python
fileobj= open("temp1.txt","r")
content= fileobj.read()
print(content)
fileobj.close()
```

```python
listOne=["one\n","two\n","three\n","four\n","five\n"]
fileobj= open("temp5.txt","w")
fileobj.writelines(listOne)    # to write multiplefile strings to the line
fileobj.close()
```

```python
fileobj = open("temp5.txt","r")
# info = fileobj.read()
# info = fileobj.readline()
info = fileobj.readlines()
print(info)
fileobj.close()
```

WITH KEY-WORDS    ...here we don't need to use close function.   

```python
with open("temp2.txt","r") as fileobj:

    info = fileobj.read()

    print(info)
```
