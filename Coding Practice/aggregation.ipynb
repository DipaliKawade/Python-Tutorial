# method overridding / polymorphism
same method name in diff classes 
The same method name pay() behaves differently depending on the object:

```python
class upi:
    def pay(self):
        print("pay using upi")
class netbanking(upi):
    def pay(self):
        print("pay using netbanking")
class creditcard(netbanking):
    def pay(self):
        print("pay using creditcard")

obj1=upi()
obj1.pay()

obj2=netbanking()
obj2.pay()

obj3=creditcard()
obj3.pay()
```

# method overloading

Method overloading means having multiple methods with the same name but different parameters.
However, Python does not support traditional method overloading like Java. We can achieve similar behavior using default arguments or *args.

```python
class calcy:
    def add(self,num1=0, num2=0,num3=0):
        print(num1+num2+num3)
obj=calcy()
obj.add(10,20,30)
obj.add(30,10,60)
obj.add(50,20,90)
```

# encapsulation
Encapsulation means wrapping data and methods together inside a class and protecting the data from direct access.

```python
class bank:
    def __init__(self):
        self.name = "dipali"
        self.__pin = 123456

    def get_pin(self):
        return self.__pin
    
    def set_pin(self,newpin):
        strnum = str(newpin)
        if len(strnum) == 6:
            self.__pin = newpin
        else:
            print("enter 6 digit pin")


class subbranch(bank):
    def display(self):
        print(self.name)
        # print(self.__pin)
        print(self.get_pin())
        self.set_pin(123476)
        print(self.get_pin())


obj=subbranch()
obj.display()

```

# abstraction

```python
class bank:

    def __init__(self):
        self.balance = 0

    def menu(self):
        print("1. Deposite")
        print("2. Withdraw")
        print("3. Exit")

        choice = int(input("enter your choice: "))

        match choice:
            case 1:
                self.deposite()
            case 2:
                self.withdraw()
            case 3:
                self.exit()

    def deposite(self):
        pass

    def withdraw(self):
        pass

    def exit(self):
        print("Thank you")


class SBI(bank):

    def deposite(self):
        amount = int(input("enter amount for deposite: "))
        self.balance += amount
        print("current balance is:", self.balance)

    def withdraw(self):
        amount = int(input("enter amount for withdraw: "))

        if amount <= self.balance:
            self.balance -= amount
            print("current balance is:", self.balance)
        else:
            print("insufficient balance")


obj = SBI()
obj.menu()
```

# use of static variable
It is common for all objects.

```python
class Student:

    college = "PES Modern College"   # static variable
    university="sppu"
    def __init__(self, name):
        self.name = name

    def display(self):
        print("")
        print("Student name:", self.name)
        print("College name:", Student.college)
        print("university:",Student.university)


obj1 = Student("Dipali")
obj2 = Student("Shubhangi")

obj1.display()
obj2.display()
```

# An instance variable is a variable that has a different value for each object.

```python
class Student:

    def __init__(self, name, rollno):
        self.name = name
        self.rollno = rollno

    def display(self):
        print("Name:", self.name)
        print("Roll No:", self.rollno)


obj1 = Student("Dipali", 101)
obj2 = Student("Shubhangi", 102)

obj1.display()
obj2.display()
```

1) . Instance Method

An instance method uses self and works with object data.

```python
class Student:

    def __init__(self, name):
        self.name = name

    def display(self):       # Instance method
        print("Student name:", self.name)


obj = Student("Dipali")
obj.display()
```

👍 Instance method can be accessed using an object, but normally not directly using the class.

```python
class Student:

    def display(self):
        print("Student details")


obj = Student()

obj.display()       # Using object ✅
Student.display(obj) # Using class ✅
```

2) . Static Method

A static method does not use self. It performs a task that does not depend on object data.

```python
class Student:

    @staticmethod       #decorator
    def college():      # college() is a static method because it does not use self.
        print("PES Modern College")

Student.college()
```

✅ A static method can be accessed using both the class and an object.

```python
class Student:

    @staticmethod
    def college():
        print("PES Modern College")


# Using class
Student.college()

# Using object
obj = Student()
obj.college()
```
